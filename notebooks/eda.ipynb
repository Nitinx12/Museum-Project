## MongoDB Exploratory Data Analysis (EDA)
    Purpose: This notebook performs Exploratory Data Analysis (EDA) on data stored in MongoDB.

```python
import sys
import os
import pandas as pd
import numpy as np
from matplotlib import pyplot as plt
import seaborn as sns

from pprint import pprint

# Add project root to Python path
sys.path.append(os.path.abspath(".."))

from utils.connection import get_mongo_db
```

```python
db = get_mongo_db()

print("Connected to MongoDB")
print("Database:", db.name)
```

```python
collections = db.list_collection_names()
print(collections)
```

```python
collection_stats = []

for collection_name in collections:
    count = db[collection_name].count_documents({})
    
    collection_stats.append({
        "collection": collection_name,
        "documents": count
    })

collection_stats = pd.DataFrame(collection_stats)

collection_stats
```
